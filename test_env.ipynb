```python
import torch
assert torch.cuda.is_available(), f"CUDA is not available."
if torch.cuda.is_available():
    gpu_properties = torch.cuda.get_device_properties(0)
    total_memory_gb = gpu_properties.total_memory / (1024**3)
    assert total_memory_gb >= 12, f"Insufficient GPU memory: {total_memory_gb:.2f} GB detected."
    print(f"CUDA is available. GPU Memory: {total_memory_gb:.2f} GB")
    if total_memory_gb >= 44:
        print("You have enough GPU memory to double the size of the images and spectral basis's recovered. Also, you can swap the medium network size with the large one.")
```
